## Key Findings

- Item-Based Collaborative Filtering achieved the best recommendation ranking performance with MAP@10 = 0.0189.
- SVD achieved the best rating prediction accuracy with RMSE = 0.9864.
- Recommendation ranking quality and rating prediction accuracy do not necessarily correlate.
- Item-Based Collaborative Filtering was selected as the preferred model for content discovery due to its superior Top-K recommendation performance.

## Model Comparison Discussion

The two recommendation approaches exhibited different strengths. Item-Based Collaborative Filtering achieved a higher MAP@10 score, indicating superior recommendation ranking quality. By leveraging similarities between movies, the model was better able to place relevant content within the user's top recommendations.

In contrast, the SVD model achieved the lowest RMSE, demonstrating stronger rating prediction accuracy. However, despite accurately estimating ratings, the model struggled to rank relevant items near the top of recommendation lists, resulting in a significantly lower MAP@10 score.

These results highlight an important distinction between rating prediction and recommendation quality. A model that predicts ratings accurately does not necessarily generate the best Top-K recommendations. For content discovery applications, ranking quality may be more important than rating prediction accuracy, making Item-Based Collaborative Filtering the preferred approach in this project.


## Success Case

Item-Based Collaborative Filtering successfully generated recommendations that aligned with user viewing behavior. By leveraging similarities between movies based on historical interaction patterns, the model was able to identify content likely to be relevant to users.


This demonstrates the strength of collaborative filtering approaches in content discovery tasks, where the objective is to recommend unseen items that match a user's interests.



## Explainable Recommendations

One advantage of Item-Based Collaborative Filtering is that recommendations can be explained using movie similarity relationships.

For example, if a user highly rated movies such as *Reservoir Dogs*, *Kill Bill: Vol. 2*, and *Seven Samurai*, the system recommends movies that were frequently liked by users who enjoyed similar content.

This makes recommendations more interpretable than latent-factor approaches such as SVD, where recommendations are generated from learned latent representations that are difficult to explain directly.

For User 562621, the recommendation system identified movies that share viewing patterns with content the user previously rated highly. Because the user showed preference for highly-rated drama and cult-classic films such as *Reservoir Dogs*, *Kill Bill: Vol. 2*, and *Seven Samurai*, the model recommended movies that similar users also tended to enjoy.

This provides a simple explanation for the recommendation process and improves transparency of the system.



## Failure Case 1: Cold-Start User (Item-Based Collaborative Filtering)

A user with very few historical ratings was selected to evaluate recommendation quality under sparse interaction conditions. Because collaborative filtering relies heavily on past user behavior, the model had insufficient information to accurately infer preferences.

As a result, recommendations became less personalized and were largely influenced by globally popular interaction patterns. This demonstrates the cold-start problem, one of the most common challenges in recommendation systems.


## Cold Start Strategy

Recommendation systems face challenges when insufficient interaction history is available.

### New Users

For users with very few ratings, collaborative filtering models cannot accurately infer preferences. A practical solution is to recommend popular or highly-rated movies until sufficient user interaction data has been collected.

Another approach is to collect initial preferences through onboarding questionnaires, allowing recommendations to be personalized from the beginning.

### New Movies

New movies have limited interaction data and therefore cannot easily participate in collaborative filtering recommendations.

To address this issue, content-based features such as genre, release year, cast information, and movie descriptions can be used to recommend similar content until enough ratings are collected.

### Sparse Interaction Histories

For users who rate very few movies, recommendation quality may decline due to limited preference information. Hybrid recommendation systems that combine collaborative filtering and content-based methods can help improve performance in such scenarios.


## Failure Case 2: Popularity Bias (Item-Based Collaborative Filtering)

Collaborative filtering models learn from historical user interactions. Movies with a large number of ratings contribute more strongly to similarity calculations and therefore appear more frequently in recommendation lists.

This can result in popularity bias, where highly-rated mainstream content is repeatedly recommended while niche or less frequently rated movies receive limited exposure.


## Failure Case: SVD Recommendation Ranking

To evaluate recommendation quality beyond RMSE, an individual user recommendation analysis was performed.

For User 1747125, the test set contained one relevant movie (Movie 143), defined as a movie with an actual rating of at least 4. However, none of the Top-10 recommendations generated by the SVD model contained this movie. As a result, the overlap between relevant and recommended movies was empty, producing an AP@10 score of 0.

This example illustrates a key limitation of the SVD model in this project. Although SVD achieved strong rating prediction accuracy (low RMSE), it struggled to rank relevant movies near the top of recommendation lists. Consequently, recommendation ranking performance (MAP@10) remained substantially lower than that of Item-Based Collaborative Filtering.

The result highlights an important distinction in recommendation systems: accurate rating prediction does not necessarily imply effective recommendation ranking.


## Cold Start Strategy

Cold-start scenarios occur when new users or new movies have insufficient interaction history for collaborative filtering models to learn meaningful patterns.

For new users, an effective strategy is to recommend popular or highly-rated movies until sufficient interaction data is collected. Another approach is to collect initial preference information through onboarding questionnaires.

For new movies, metadata such as genre, release year, cast information, or content descriptions can be incorporated through content-based recommendation methods. Hybrid recommendation systems that combine collaborative and content-based approaches are commonly used to address cold-start challenges in production environments.


## Scalability Analysis

Item-Based Collaborative Filtering requires computation and storage of a movie similarity matrix. As the number of items increases, both memory requirements and similarity computation costs grow substantially.

SVD is generally more scalable for large recommendation systems because it compresses user-item interactions into a lower-dimensional latent factor representation. Once trained, prediction generation is efficient and memory requirements remain relatively modest.

For very large-scale streaming platforms, matrix factorization approaches are typically preferred due to their scalability advantages.


## Conclusion

Two recommendation approaches were evaluated on the Netflix Prize Dataset: Item-Based Collaborative Filtering and Singular Value Decomposition (SVD).

The SVD model achieved the strongest rating prediction performance, producing the lowest RMSE score. However, Item-Based Collaborative Filtering significantly outperformed SVD in recommendation ranking quality, achieving a substantially higher MAP@10 score.

Based on both quantitative evaluation metrics and qualitative analysis, Item-Based Collaborative Filtering was selected as the preferred recommendation approach for personalized content discovery. The project demonstrates the importance of evaluating recommendation systems using both prediction accuracy and ranking quality metrics.
