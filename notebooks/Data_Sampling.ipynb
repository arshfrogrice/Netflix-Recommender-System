```python
import pandas as pd
from tqdm import tqdm

def sample_movies(files, max_movies=1000):

    data = []

    current_movie = None
    movie_counter = 0

    for filepath in files:

        print(f"Reading {filepath}")

        with open(filepath, "r", encoding="utf-8") as f:

            for line in tqdm(f):

                if line.endswith(":\n"):

                    current_movie = int(line[:-2])

                    movie_counter += 1

                    if movie_counter > max_movies:

                        return pd.DataFrame(
                            data,
                            columns=[
                                "user_id",
                                "movie_id",
                                "rating",
                                "date"
                            ]
                        )

                else:

                    user_id, rating, date = (
                        line.strip().split(",")
                    )

                    data.append([
                        int(user_id),
                        current_movie,
                        int(rating),
                        date
                    ])

    return pd.DataFrame(
        data,
        columns=[
            "user_id",
            "movie_id",
            "rating",
            "date"
        ]
    )
```

```python
files = [
    "../data/combined_data_1.txt",
    "../data/combined_data_2.txt",
    "../data/combined_data_3.txt",
    "../data/combined_data_4.txt"
]

df_movies = sample_movies(
    files,
    max_movies=1000
)
```

```python
print(df_movies.shape)
print(df_movies.movie_id.nunique())
print(df_movies.user_id.nunique())
```
